```python
from google import genai
from google.genai import types
```

```python
#tool call
def get_weather(city:str)->dict:
    """
    Fetches the current real-time weather details for a specific city.

    use this tool when user asks about the current weather ,temperature
    or climate condition of a location.Don't use for the future forcasts.

    Args:
        city:The name of the city user look up(e.g , "London","Dhaka","tokyo")

    Returns:
        A dictionary containning the keys:
        "temparature":the current temperature in celsius(float).
        "condition":the current weather state,such as "sunny","raining" (str)
    
    """
    return {"temperature":30,"condition":"sunny"}
```

```python
search_tool=types.Tool(
    google_search=types.GoogleSearch()
)
```

```python
client=genai.Client()
response=client.models.generate_content(
    model="gemini-3.1-flash-lite",
    contents="What is the current fashion trend in New York right now?",
    config=types.GenerateContentConfig(
        tools=[get_weather,search_tool],
        temperature=0.7
    )
)

print(response.text)
```
